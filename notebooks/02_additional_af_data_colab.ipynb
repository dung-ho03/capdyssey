# 추가 손목 AF·비AF 자료 확보

39명 장기 기록의 v3이며 비영리 연구용 CC BY-NC-SA 4.0 자료입니다. 기본 2명 묶음은 약 1.42GB입니다. 이 노트북은 다운로드·무결성 검사만 수행하며 기존 CNN에 바로 입력하지 않습니다. ECG 주석과 PPG 시간 정렬·참가자 분할을 먼저 완료해야 합니다.

[이용 조건과 확장 이유](https://github.com/dung-ho03/capdyssey/blob/main/docs/AF_DATA_EXPANSION.md)

```python
from pathlib import Path
import subprocess, sys
REPO = Path('/content/capdyssey_af_data')
if not REPO.exists():
    subprocess.run(['git','clone','https://github.com/dung-ho03/capdyssey.git',str(REPO)],check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','-r',str(REPO / 'requirements-data.txt')],check=True)
```

```python
SUBJECTS = ['006', '039']
subprocess.run([sys.executable,'-u',str(REPO / 'scripts/download_longterm_af.py'),'--subjects',*SUBJECTS],check=True)
```

```python
import json
folder = REPO / 'data/raw/longterm_af_v3'
print(json.dumps(json.loads((folder / 'download-status.json').read_text()),indent=2))
```

```python
subprocess.run([sys.executable, str(REPO / 'scripts/audit_longterm_af.py'), '--subjects', *SUBJECTS, '--output', str(folder / 'structure-audit.json')], check=True)
print((folder / 'structure-audit.json').read_text())
```
