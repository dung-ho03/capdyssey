# 세 자료 10초 잠정 실험
손목 0=비AF, 1=AF는 잠정 해석. 0.25/0.5와 경계 구간 제외. 공식 확인 또는 임상 검증 결과가 아닙니다. 원본 자료와 기존 y=-1은 보존합니다.
실험 이유와 제한: docs/AF_PROVISIONAL_EXPERIMENT.md. 실험 결과는 Drive에 저장됩니다.

```
from google.colab import drive
from pathlib import Path
import subprocess, sys
drive.mount('/content/drive')
repo = Path('/content/capdyssey_provisional')
if not repo.exists():
    subprocess.run(['git','clone','https://github.com/dung-ho03/capdyssey.git',str(repo)],check=True)
subprocess.run([sys.executable,'-m','unittest','discover','-s','tests','-p','test_provisional_labels.py'],cwd=repo,check=True)
base = Path('/content/drive/MyDrive/Capdyssey/af_integrated')
bundle = base/'processed/provisional_exact01_v1'
output = base/'models/provisional_exact01_v1'
```

```
if not (bundle/'manifest.json').exists():
    subprocess.run([sys.executable,'scripts/prepare_af_provisional.py','--bundle',str(base/'processed/integrated_10s_resume_v1'),'--wrist',str(base/'raw/longterm_af_v3'),'--output',str(bundle)],cwd=repo,check=True)
else:
    print('REUSING COMPLETED PROVISIONAL BUNDLE',bundle)
```

```
subprocess.run([sys.executable,'-u','scripts/train_af_integrated.py','--bundle',str(bundle),'--output',str(output),'--epochs','20','--provisional'],cwd=repo,check=True)
```
